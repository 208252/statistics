##### **Task 1:** Replace `coordinates` with `latitude` and `longitude` columns

To perform this task, we will use the following methods:

- `.str.replace("","")` replaces one string in each row of a column with another
- `.str.split("")` takes in a string and lets you split a column into two based on that string
- `.astype()` lets you convert a column from one type to another

````python
airbnb[['latitude', 'longitude']] = (
    airbnb['coordinates']
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
    .str.split(',', n=1, expand=True)
    .astype(float)
)

airbnb = airbnb.drop(columns=['coordinates'])

airbnb.head(5)
````

##### **Task 2:** Remove `$` from `price` and convert it to `float`

To perform this task, we will be using the following methods:

- `.str.strip()` which removes a specified string from each row in a column
- `.astype()`

````python
airbnb['price'] = airbnb['price'].astype('string')
airbnb['price'] = airbnb['price'].str.strip("$")
airbnb['price'] = airbnb['price'].astype('float')

airbnb['price'].head()
````

##### **Task 3:** Convert `listing_added` and `last_review` columns to `datetime`

````python
airbnb['listing_added'] = pd.to_datetime(airbnb['listing_added'], format='%Y-%m-%d')

airbnb['last_review'] = pd.to_datetime(airbnb['last_review'], format='%Y-%m-%d')
print(airbnb.dtypes)
````

##### **Task 9:** Let's deal with duplicate data


There are two notable types of duplicate data:

- Identical duplicate data across all columns
- Identical duplicate data cross most or some columns

To diagnose, and deal with duplicate data, we will be using the following methods and functions:

- `.duplicated(subset = , keep = )`
  - `subset` lets us pick one or more columns with duplicate values.
  - `keep` returns lets us return all instances of duplicate values.
- `.drop_duplicates(subset = , keep = )`
  

````python
# Print the header of the DataFrame again
airbnb.head()
````

````python
# Find duplicates
airbnb[airbnb.duplicated()]

````

````python
# Remove identical duplicates
airbnb = airbnb.drop_duplicates()
print(airbnb.shape)
````

````python
# Find non-identical duplicates
airbnb[airbnb.duplicated(subset='listing_id', keep=False)]
````

````python
# Show all duplicates
airbnb[airbnb.duplicated(subset='listing_id', keep=False)].sort_values('listing_id')
````

To treat identical duplicates across some columns, we will chain the `.groupby()` and `.agg()` methods where we group by the column used to find duplicates (`listing_id`) and aggregate across statistical measures for `price`, `rating` and `list_added`. The `.agg()` method takes in a dictionary with each column's aggregation method - we will use the following aggregations:

- `mean` for `price` and `rating` columns
- `max` for `listing_added` column
- `first` for all remaining column

*A note on dictionary comprehensions:*

Dictionaries are useful data structures in Python with the following format
`my_dictionary = {key: value}` where a `key` is mapped to a `value` and whose `value` can be returned with `my_dictionary[key]` - dictionary comprehensions allow us to programmatically create dicitonaries using the structure:

```
{x: x*2 for x in [1,2,3,4,5]}
{1:2, 2:4, 3:6, 4:8, 5:10}
```

````python
cols = [c for c in airbnb.columns if c != 'listing_id']

agg_dict = {
    c: 'mean' if c in ['price', 'rating']
    else 'max' if c == 'listing_added'
    else 'first'
    for c in cols
}

airbnb = airbnb.groupby('listing_id').agg(agg_dict).reset_index()
print(airbnb.shape)
airbnb.head()
````

````python
#data cleansing, setting proper min price
airbnb = airbnb[airbnb['price'] > 0]
print(f"New min price: {airbnb['price'].min()}")
````

# Cross-sectional analysis

### Price comparison by borough

````python

airbnb['price'] = airbnb['price'].astype(str).str.strip("$")
airbnb['price'] = pd.to_numeric(airbnb['price'], errors='coerce')


if 'borough' not in airbnb.columns and 'neighbourhood_full' in airbnb.columns:
    airbnb['borough'] = airbnb['neighbourhood_full'].str.split(',').str[0]

price_by_borough = airbnb.groupby('borough')['price'].median().sort_values(ascending=False)
print("Median Price by Borough:\n", price_by_borough)

plt.figure(figsize=(10, 6))
sns.boxplot(x='borough', y='price', data=airbnb)
plt.title('Cross-Sectional Analysis: Airbnb Prices by Borough')
plt.ylim(0, 500)
plt.show()
````

### Room type vs Rating Quality Analysis

````python

rating_by_room = airbnb.groupby('room_type')['rating'].mean().sort_values(ascending=False)
print("Average Rating by Room Type:\n", rating_by_room)

plt.figure(figsize=(8, 5))
sns.violinplot(x='room_type', y='rating', data=airbnb)
plt.title('Cross-Sectional Analysis: Ratings across Room Types')
plt.show()
````

### Price & Quality Intersection, comparing the prices of highly-rated listings versus lower-rated listings

````python

airbnb['rating_tier'] = np.where(airbnb['rating'] >= 4.5, 'High Rated (>=4.5)', 'Standard (<4.5)')

tier_comparison = airbnb.groupby('rating_tier')['price'].mean()
print("Average Price by Rating Tier:\n", tier_comparison)


plt.figure(figsize=(6, 4))
sns.barplot(x='rating_tier', y='price', data=airbnb, estimator=np.mean)
plt.title('Cross-Sectional Analysis: Price Premium for Highly Rated Airbnbs')
plt.ylabel('Average Price ($)')
plt.show()
````

````python
from IPython.display import display, Markdown
x = airbnb[airbnb['price'] <= 400]['price']
mean_val = x.mean()
median_val = x.median()

quantitative = airbnb[['price', 'rating', 'number_of_reviews', 'availability_365']]

def markdown_summary_advanced(df, round_decimals=3):
    summary = df.describe().T  
    summary['IQR'] = summary['75%'] - summary['25%']
    summary['Mode'] = df.mode().iloc[0].values
    summary['CV'] = summary['std'] / summary['mean']
    Q1 = df.quantile(0.25)
    Q3 = df.quantile(0.75)
    IQR_calc = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR_calc
    upper_bound = Q3 + 1.5 * IQR_calc
    summary['Outliers'] = ((df < lower_bound) | (df > upper_bound)).sum().values
    summary['Skewness'] = df.skew()
    summary['Kurtosis'] = df.kurt()
    summary = summary.round(round_decimals)
    
    headers = [''] + list(summary.columns)
    lines = []
    lines.append('| ' + ' | '.join(headers) + ' |')
    lines.append('| ' + ' | '.join(['---'] * len(headers)) + ' |')
    for idx, row in summary.iterrows():
        row_vals = [str(idx)] + [str(val) for val in row]
        lines.append('| ' + ' | '.join(row_vals) + ' |')
    return '\n'.join(lines)


display(Markdown(markdown_summary_advanced(quantitative)))

fig, (ax_box, ax_hist) = plt.subplots(
    2, 
    sharex=True, 
    gridspec_kw={"height_ratios": (.15, .85)}, 
    figsize=(12, 8)
)

sns.boxplot(x=airbnb[airbnb['price'] <= 400]['price'], ax=ax_box, color='skyblue')
ax_box.set(xlabel='')
ax_box.set_title('Composite Univariate Diagnostic: Distribution of Airbnb Prices', fontsize=16, fontweight='bold', pad=15)

sns.histplot(x=airbnb[airbnb['price'] <= 400]['price'], ax=ax_hist, kde=True, color='teal', stat='density')

ax_hist.axvline(mean_val, color='crimson', linestyle='--', linewidth=2.5, label=f'Arithmetic Mean (${mean_val:.1f})')
ax_hist.axvline(median_val, color='orange', linestyle='-', linewidth=2.5, label=f'Median (${median_val:.1f})')
ax_hist.legend(fontsize=11)

plt.xlabel('Price ($)', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

### **1. Correlation Analysis (Continuous Variables)**

We compared **Pearson ($r$)** (straight-line relationship), **Kendall's Tau ($\tau$)** (rank agreement), and **Spearman ($\rho$)** (rank curves) between our numerical variables.

#### **Takeaways:**
- **Price vs. Rating ($r \approx -0.005$):** 
  * **Takeaway:** Absolutely **zero correlation**! Paying more does not get you a better rating. A cheap private room is just as likely to have a 5-star rating as a luxury penthouse.
- **Availability vs. Reviews (Spearman $\rho \approx 0.226$ vs. Pearson $r \approx 0.169$):**
  * **Takeaway:** A weak but real positive trend. Listings open more days of the year accumulate more reviews.
  * **Stats Tip:** Spearman is higher than Pearson because this relationship is a curve, not a straight line. Spearman is much better at picking up non-linear trends!
- **Price vs. Availability ($r \approx 0.086$):**
  * **Takeaway:** Almost no correlation. Being open more days doesn't let hosts charge significantly more.

````python

contingency_table = pd.crosstab(airbnb['room_type'], airbnb['borough'])
print("Contingency Table (Room Type vs. Borough):")
display(contingency_table)

chi2, p_val, dof, expected = ss.chi2_contingency(contingency_table)
print("\n--- Chi-Square Test of Independence ---")
print(f"Chi-Square Statistic : {chi2:.4f}")
print(f"p-value              : {p_val:.4e}")
print(f"Degrees of Freedom   : {dof}")

cramers_v = ss.contingency.association(contingency_table, method="cramer")
print(f"Cramer's V           : {cramers_v:.4f}")
````

### **2. Association Analysis (Categorical Variables)**

We ran a **Chi-Square Test** and calculated **Cramer's V** to see if listing location (`borough`) is related to the choice of room offered (`room_type`).

- **Hypotheses:** 
  * $H_0$: Borough and room type are independent.
  * $H_a$: Borough and room type are dependent (related).

#### **Takeaways:**
- **p-value ($1.947 \times 10^{-83}$):** Extremely small! We reject $H_0$. Location and room type are definitely related.
- **Cramer's V ($0.1432$):** A value of ~0.14 shows a **weak-to-moderate** relationship. It is statistically real, but location isn't the *only* thing driving room type.
- **Geographical Vibe:**
  * **Manhattan:** Heavy on **Entire Apartments** (privacy-focused, high budget).
  * **Brooklyn:** Almost a 50/50 split between apartments and private rooms.
  * **Queens:** Heavy on **Private Rooms** (saving money, sharing space).

````python

df_filtered = airbnb[airbnb['price'] <= 400]

fig, axes = plt.subplots(1, 2, figsize=(20, 8))

sns.boxplot(data=df_filtered, x='room_type', y='price', palette='Set2', hue='room_type', ax=axes[0], legend=False)
axes[0].set_title('Price Distribution by Room Type (Price <= $400)', fontsize=14, fontweight='bold', pad=10)
axes[0].set_xlabel('Room Type', fontsize=12)
axes[0].set_ylabel('Price ($)', fontsize=12)
axes[0].grid(True, linestyle='--', alpha=0.3)

sns.boxplot(data=df_filtered, x='borough', y='price', palette='Set3', hue='borough', ax=axes[1], legend=False)
axes[1].set_title('Price Distribution by Borough (Price <= $400)', fontsize=14, fontweight='bold', pad=10)
axes[1].set_xlabel('Borough', fontsize=12)
axes[1].set_ylabel('Price ($)', fontsize=12)
axes[1].grid(True, linestyle='--', alpha=0.3)

plt.suptitle('Quantitative vs. Qualitative Price Comparisons', fontsize=16, fontweight='bold', y=0.98)
plt.tight_layout()
plt.show()

groups_rt = [group['price'].values for name, group in df_filtered.groupby('room_type')]
kw_stat_rt, kw_p_rt = ss.kruskal(*groups_rt)
print("--- Kruskal-Wallis H-Test: Price across Room Types ---")
print(f"H-Statistic : {kw_stat_rt:.4f}")
print(f"p-value     : {kw_p_rt:.4e}")

groups_bg = [group['price'].values for name, group in df_filtered.groupby('borough')]
kw_stat_bg, kw_p_bg = ss.kruskal(*groups_bg)
print("\n--- Kruskal-Wallis H-Test: Price across Boroughs ---")
print(f"H-Statistic : {kw_stat_bg:.4f}")
print(f"p-value     : {kw_p_bg:.4e}")
````

### **3. Quantitative vs. Qualitative Analysis (Continuous vs. Categorical Variables)**

We compared **Price** (filtered to $\le \$400$ to control for extreme outliers) across **Room Type** and **Borough**.

* **Stats Tip:** Because prices are heavily skewed and not normally distributed, we used the non-parametric **Kruskal-Wallis Test** instead of standard ANOVA.

#### **Takeaways:**
1. **Price by Room Type ($p < 2.2 \times 10^{-16}$):**
   * **Takeaway:** Massive differences! Medians are extremely different ($H \approx 4495.73$). Renting an entire apartment is significantly more expensive than a private or shared room.
2. **Price by Borough ($p \approx 2.13 \times 10^{-307}$):**
   * **Takeaway:** Huge location premium ($H \approx 1425.42$). Manhattan has the highest median price (\$150), and Bronx has the lowest (\$65).
3. **Price vs. Ratings:**
   * **Takeaway:** Location and room type heavily drive price, but listing ratings stay the same across all of them ($p \approx 0.788$). Guests adjust their expectations based on what they pay!
